```python
import re

role_mapping = {
    1: ["chief", "ceo", "cfo", "coo", "cio", "cto", "chief executive", "chief officer", "president", "vice president", "vp", "director"],
    0: []
}

# regex patterns
compiled_roles = {
    role_id: re.compile(r'\b(?:' + '|'.join([re.escape(k) for k in keywords]) + r')\b', re.IGNORECASE)
    for role_id, keywords in role_mapping.items() if keywords
}

# categorize role
def get_role_label(row):
    combined_text = ' '.join([
        str(row.get('Intro', '')),
        str(row.get('Experiences', ''))
    ]).lower()

    for role_id in sorted(compiled_roles.keys(), reverse=True):  # Check higher positions first
        if compiled_roles[role_id].search(combined_text):
            return role_id
    return 0

df['Role_Level'] = df.apply(get_role_label, axis=1)

print(df[['Intro', 'Experiences', 'Role_Level']].head())

label_counts = df['Role_Level'].value_counts().sort_index()

label_description = {
    0: "Non high level position",
    1: "High level position"
}

for label, count in label_counts.items():
    print(f"Label {label} ({label_description[label]}): {count} profiles")


```

```python
from transformers import DistilBertTokenizer, DistilBertModel
import torch
import numpy as np

tokenizer = DistilBertTokenizer.from_pretrained('distilbert-base-uncased')
model = DistilBertModel.from_pretrained('distilbert-base-uncased')

def get_bert_embedding(text):
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding=True, max_length=128)
    with torch.no_grad():
        outputs = model(**inputs)
    return outputs.last_hidden_state[:, 0, :].squeeze().numpy()
```

```python
df['About'] = df['About'].fillna('')
df['About_Embedding'] = df['About'].apply(get_bert_embedding)
```

```python
# Combine numeric and BERT features
numeric_columns = [col for col in df.columns if df[col].dtype in ['int64', 'float64'] and col != 'Label' and col != "Role_Level"]
about_embeddings = np.stack(df['About_Embedding'].values)
numeric_features = df[numeric_columns].values
X_combined = np.concatenate([numeric_features, about_embeddings], axis=1)
y = df['Role_Level']
```

```python
from sklearn.decomposition import PCA

about_embeddings = np.stack(df['About_Embedding'].values)

pca = PCA(n_components=1)
about_reduced = pca.fit_transform(about_embeddings)

df['About'] = about_reduced[:, 0]

df = df.drop("Unnamed: 0", axis=1)
df = df.drop("Label", axis=1)

numeric_columns = [col for col in df.columns if df[col].dtype in ['int64', 'float64'] and col != 'Role_Level']
X_combined = df[numeric_columns + ['About']].values  # or include PC2 as well
y = df['Role_Level']
```

```python
# Random Forest
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV

X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.3, random_state=42)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

rf_model = RandomForestClassifier(random_state=42)

param_grid_rf = {
    'n_estimators': [50, 100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'max_features': ['sqrt', 'log2']
}

grid_search_rf = GridSearchCV(estimator=rf_model, param_grid=param_grid_rf, scoring='accuracy', cv=5, n_jobs=-1)

print("Starting GridSearchCV for RandomForestClassifier...")
grid_search_rf.fit(X_train_scaled, y_train)

best_rf_model = grid_search_rf.best_estimator_

print("\nBest parameters found for RandomForestClassifier: ", grid_search_rf.best_params_)

y_pred_test_tuned_rf = best_rf_model.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print(classification_report(y_test, y_pred_test_tuned_rf, digits=3))
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
from sklearn.pipeline import Pipeline

X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.3, random_state=42)

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('lr', LogisticRegression(max_iter=1000, random_state=42))
])

param_grid = {
    'lr__C': [0.01, 0.1, 1, 10, 100],
    'lr__penalty': ['l1', 'l2'],
    'lr__solver': ['liblinear']
    # If you want to try 'lbfgs', 'newton-cg', 'sag', 'saga', you'd typically use 'l2' or 'none' penalties.
    # e.g., {'lr__penalty': ['l2'], 'lr__solver': ['lbfgs', 'newton-cg', 'sag', 'saga']}
}

grid_search_lr = GridSearchCV(estimator=pipeline, param_grid=param_grid, scoring='accuracy', cv=5, n_jobs=-1)

grid_search_lr.fit(X_train, y_train)

best_lr_model = grid_search_lr.best_estimator_

print("\nBest parameters found for Logistic Regression: ", grid_search_lr.best_params_)

y_pred_test_tuned_lr = best_lr_model.predict(X_test)
```

```python
print(classification_report(y_test, y_pred_test_tuned_lr, digits=3))
```

```python
# Naive Bayes
# Naive Bayes
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report

X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.3, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

nb_model = GaussianNB()

param_grid_nb = {
    'var_smoothing': [1e-9, 1e-8, 1e-7, 1e-6]
}

grid_search_nb = GridSearchCV(estimator=nb_model, param_grid=param_grid_nb, scoring='accuracy', cv=5, n_jobs=-1)

grid_search_nb.fit(X_train_scaled, y_train)

best_nb_model = grid_search_nb.best_estimator_

print("\nBest parameters found for Gaussian Naive Bayes: ", grid_search_nb.best_params_)

y_pred_test_tuned_nb = best_nb_model.predict(X_test_scaled)
```

```python
print(classification_report(y_test, y_pred_test_tuned_nb, digits=3))
```

```python
# SVC
from sklearn.svm import SVC # Import SVC
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
from sklearn.pipeline import Pipeline



# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.3, random_state=42)

pipeline_svc = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC(random_state=42))
])

param_grid_svc = {
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': ['scale', 'auto'],
    'svc__kernel': ['rbf', 'linear']
}

# Initialize GridSearchCV
grid_search_svc = GridSearchCV(estimator=pipeline_svc, param_grid=param_grid_svc, scoring='accuracy', cv=5, n_jobs=-1)

print("Starting GridSearchCV for SVC...")
grid_search_svc.fit(X_train, y_train)

best_svc_model = grid_search_svc.best_estimator_

print("\nBest parameters found for SVC: ", grid_search_svc.best_params_)

y_pred_test_tuned_svc = best_svc_model.predict(X_test)
```

```python
print(classification_report(y_test, y_pred_test_tuned_svc, digits=3))
```

```python
# Ensemble
from sklearn.ensemble import StackingClassifier

X_train, X_test, y_train, y_test = train_test_split(X_combined, y, test_size=0.3, random_state=42)

estimators = [
    ('rf', best_rf_model),
    ('lr', best_lr_model),
    ('nb', best_nb_model),
    ('svc', best_svc_model)
]

pipeline_stack = Pipeline([
    ('scaler', StandardScaler()),
    ('stack', StackingClassifier(estimators=estimators, final_estimator=LogisticRegression(random_state=42), cv=5))
])

param_grid_stack = {
    'stack__final_estimator__C': [0.01, 0.1, 1, 10, 100],
    'stack__final_estimator__penalty': ['l1', 'l2'],
    'stack__final_estimator__solver': ['liblinear']
}

grid_search_stack = GridSearchCV(estimator=pipeline_stack, param_grid=param_grid_stack, scoring='accuracy', cv=5, n_jobs=-1)

grid_search_stack.fit(X_train, y_train)
best_stack_model = grid_search_stack.best_estimator_

y_pred_test_tuned_stack = best_stack_model.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report
```

```python
# Add necessary imports if not already present
import shap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt # Although not plotting in this block, keep for consistency
from IPython.display import display # Keep display for potential DataFrame output
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords if not already
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')


# --- Reload original data for text ---
# Replicate data loading and preprocessing steps before PCA and Intro drop
# This is necessary to get the original 'About' text aligned with the indices of X_test_df
initial_linkedinData = pd.read_pickle("LinkedIn_Dataset.pcl")
initial_linkedinData = initial_linkedinData[initial_linkedinData["Label"] == 0]
initial_linkedinData.reset_index(drop=True, inplace=True)
df_before_intro_drop = pd.read_csv('Updated_LinkedIn_Dataset.csv', na_values=["NaN", "nan", "NAN"])
df_before_intro_drop = df_before_intro_drop.dropna()

# Re-apply Role_Level mapping for correct undersampling (ensure consistent row selection)
role_mapping = {
    1: ["chief", "ceo", "cfo", "coo", "cio", "cto", "chief executive", "chief officer", "president", "vice president", "vp",  "director"],
    0: []
}
compiled_roles = {
    role_id: re.compile(r'\b(?:' + '|'.join([re.escape(k) for k in keywords]) + r')\b', re.IGNORECASE)
    for role_id, keywords in role_mapping.items() if keywords
}
def get_role_label(row):
    combined_text = ' '.join([
        str(row.get('Intro', '')),
        str(row.get('Experiences', ''))
    ]).lower()
    for role_id in sorted(compiled_roles.keys(), reverse=True):
        if compiled_roles[role_id].search(combined_text):
            return role_id
    return 0
df_before_intro_drop['Role_Level'] = df_before_intro_drop.apply(get_role_label, axis=1)

# Apply undersampling exactly as done for 'df'
from sklearn.utils import resample
label_0 = df_before_intro_drop[df_before_intro_drop['Role_Level'] == 0]
label_1 = df_before_intro_drop[df_before_intro_drop['Role_Level'] == 1]
label_0_undersampled = resample(label_0, replace=False, n_samples=454, random_state=42)
label_1_undersampled = resample(label_1, replace=False, n_samples=454, random_state=42)
df_with_original_text = pd.concat([label_0_undersampled, label_1_undersampled])
df_with_original_text = df_with_original_text.sample(frac=1, random_state=42).reset_index(drop=True)


# Assume explainer, X_test, y_test, X_test_df, shap_values, combined_feature_names, class_labels are defined
# Make sure these variables from your previous steps are available in the environment

# --- Parameters ---
class_to_analyze = 1 # Class to analyze (0 or 1)
num_samples_to_show = 10 # <--- Changed from 5 to 10

# Find the index of the 'About' feature
# Assuming shap_values is 2D (n_samples, n_features) from shap.LinearExplainer
# Assuming X_test_scaled_df and combined_feature_names are defined
# Assuming df_with_original_text is correctly created and indexed

# Find the index of the 'About' feature
about_feature_index = combined_feature_names.index('About')

# Get the SHAP values just for the 'About' feature for ALL samples
about_shap_values = shap_values[:, about_feature_index] # Correctly slicing a 2D array

# Get the actual 'About' feature values (PC1 values)
# Assuming 'About' is a column in X_test_scaled_df (the scaled test data used for SHAP)
about_feature_values = X_test_scaled_df['About'].values # Get the PC1 values from the scaled data

# --- Find samples with the highest positive and negative SHAP values for the 'About' feature ---

# Get indices of samples in X_test (and X_test_scaled_df) sorted by About SHAP value
sorted_indices_test_set = np.argsort(about_shap_values)

# Indices for samples with highest positive About SHAP values (most increasing Class 1 prob)
num_samples_to_show = 10 # Or however many samples you want to show
top_positive_impact_indices_test_set = sorted_indices_test_set[-num_samples_to_show:]

# Indices for samples with most negative About SHAP values (most decreasing Class 1 prob)
top_negative_impact_indices_test_set = sorted_indices_test_set[:num_samples_to_show]

# Define labels for your classes for printing
class_labels = {
    0: "Class 0 (Non high level position)",
    1: "Class 1 (High level position)"
}

# --- Print the ORIGINAL 'About' text for these influential samples ---

# Print for samples increasing Class 1 probability
print(f"\n--- Top {num_samples_to_show} Original 'About' Texts where 'About' Feature STRONGLY INCREASED prediction for {class_labels.get(1, 'Unknown Class')} ---")
print(f"(Sorted by positive SHAP value for 'About' feature)")

# Sort the positive indices based on their SHAP values to print in order (highest SHAP first)
top_positive_shaps = about_shap_values[top_positive_impact_indices_test_set]
sorted_positive_test_indices = top_positive_impact_indices_test_set[np.argsort(top_positive_shaps)][::-1] # Sort descending

for i, test_idx in enumerate(sorted_positive_test_indices):
    # Get the original index from the index of X_test_scaled_df
    original_idx = X_test_scaled_df.index[test_idx]
    shap_val = about_shap_values[test_idx]
    pc1_val = about_feature_values[test_idx]
    # Use .loc with the original index to get the text from df_with_original_text
    original_text = df_with_original_text.loc[original_idx, 'About']

    print(f"\nSample {i+1} (Original Index: {original_idx}, SHAP: {shap_val:.4f}, PC1: {pc1_val:.4f}):")
    print(original_text)
    print("-" * 20) # Separator


# Print for samples decreasing Class 1 probability (increasing Class 0 probability)
print(f"\n--- Top {num_samples_to_show} Original 'About' Texts where 'About' Feature STRONGLY DECREASED prediction for {class_labels.get(1, 'Unknown Class')} (Increased prediction for {class_labels.get(0, 'Unknown Class')}) ---")
print(f"(Sorted by negative SHAP value for 'About' feature)")

# Sort the negative indices based on their SHAP values to print in order (most negative first)
top_negative_shaps = about_shap_values[top_negative_impact_indices_test_set]
sorted_negative_test_indices = top_negative_impact_indices_test_set[np.argsort(top_negative_shaps)] # Sort ascending (most negative first)

for i, test_idx in enumerate(sorted_negative_test_indices):
    # Get the original index from the index of X_test_scaled_df
    original_idx = X_test_scaled_df.index[test_idx]
    shap_val = about_shap_values[test_idx]
    pc1_val = about_feature_values[test_idx]
    # Use .loc with the original index to get the text from df_with_original_text
    original_text = df_with_original_text.loc[original_idx, 'About']

    print(f"\nSample {i+1} (Original Index: {original_idx}, SHAP: {shap_val:.4f}, PC1: {pc1_val:.4f}):")
    print(original_text)
    print("-" * 20) # Separator


print("\n--- Analysis Note ---")
print(f"Review the text content above. Look for common themes, keywords, or writing styles that appear frequently in the 'INCREASED prediction' list versus the 'DECREASEED prediction' list.")
print("This manual review helps understand what aspects of the 'About' text the model's 'About' feature (PC1) is reacting to.")
```
